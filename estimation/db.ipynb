```python
import os
import subprocess


def get_repo_root():
    try:
        repo_root = subprocess.check_output(['git', 'rev-parse', '--show-toplevel'],
                                            stderr=subprocess.STDOUT)
        return repo_root.strip().decode('utf-8')
    except subprocess.CalledProcessError as e:
        return None


repo_root = get_repo_root()
```

```python
pd.set_option('display.max_rows', 25)
pd.set_option('display.max_columns', 100)

directory_path = "estimation/tsv_export"
directory_path = os.path.join(repo_root, directory_path)
tables = {}

for file_name in os.listdir(directory_path):
    if file_name.endswith(".tsv"):  # Check if the file has a .tsv extension
        file_path = os.path.join(directory_path, file_name)  # Full path to the file
        try:
            df = pd.read_csv(file_path, sep='\t', low_memory=False)
            tables[file_name] = df
            print(f"Loaded {file_name} with shape {df.shape}")
        except Exception as e:
            print(f"Error reading {file_name}: {e}")

for file_name, df in tables.items():
    print(f"First 5 rows of {file_name}:")
    display(df.head())
```

I guess the simplest design is like this: 
For a given person (WCA ID / personId), get all of their results, and extract a distribution from which we can draw their estimated performance. 

This should have components:
1. Previous performance, mean and std dev
2. Recent performance, ie. more recent results should be weighted higher with some exponential decay according to the result date
3. Competition results, ie. results from competitions that are more competitive should be weighted higher. This can come later, but e.g. as a modifier we can see how much they over/underperform during championship competitions. 
4. Competition results, as in when all competitors in a given competition underperformed their respective means, we can infer that there were confounding factors (e.g. scramble).
5. Some modifier on whether the competitors has broken PRs consistently and recently.

Not sure what roundTypeId d, c, e mean, but "f" for finals seems consistent. 

Should probably tune this halflife somewhat toward whatever is mle for nationals

```python
def time_weighted_distribution(person_id, event_id, halflife, results, competitions):
    person_results = results[results["personId"] == person_id]
    person_results = person_results.merge(
        competitions[['id', 'name', 'year', 'month', 'day']], how='left',
        left_on='competitionId', right_on='id')

    event_results = person_results[person_results['eventId'] == event][[
        'competitionId',
        'roundTypeId',
        'best', 'average', 'personName', 'personId', 'value1', 'value2', 'value3',
        'value4',
        'value5', 'year', 'month', 'day'
    ]].assign(date=lambda x: pd.to_datetime(x[['year', 'month', 'day']]).dt.date).drop(
        columns=['year', 'month', 'day'])

    value_columns = [col for col in event_results.columns if col.startswith("value")]
    id_vars = [col for col in event_results.columns if col not in value_columns]

    event_results = pd.melt(
        event_results,
        id_vars=id_vars,
        value_vars=value_columns,
        var_name="solve_number",
        value_name="result"
    )
    event_results['date'] = pd.to_datetime(event_results['date'])
    greatest_date = event_results['date'].max()

    event_results['time_difference'] = (greatest_date.year - event_results[
        'date'].dt.year) * 12 + (greatest_date.month - event_results['date'].dt.month)
        
    dnf_rate = event_results[(event_results.time_difference <= 12) & (event_results.roundTypeId == 'f')]
    dnf_rate = dnf_rate[dnf_rate.result == -1].shape[0] / dnf_rate.shape[0]
    
    event_results = event_results[event_results.result != -1]
    event_results['weight'] = 2 ** (-event_results['time_difference'] / halflife)

    event_results.sort_values(['date'], ascending=False).head(25)

    weighted_mean = np.average(event_results['result'], weights=event_results['weight'])
    print(f"Weighted mean of results: {weighted_mean:.2f}")

    grouped = event_results.groupby(['date', 'roundTypeId', 'competitionId', 'weight'])
    group_std_devs = grouped['result'].std().reset_index(
        name='group_std_dev')

    weighted_avg_std_dev = np.average(group_std_devs['group_std_dev'],
                                      weights=group_std_devs['weight'])

    print(f"Weighted Average Standard Deviation: {weighted_avg_std_dev:.2f}")
    print(f"DNF Rate: {dnf_rate * 100:.3f}%\n")

    return weighted_mean, weighted_avg_std_dev, dnf_rate
```

```python
from scipy.stats import skewnorm, norm
def calculate_mle(data, dist, params):
    return np.sum(dist.logpdf(data, *params))
```

```python
person_id = '2018MURR03'
event = '333'
halflife = 3

person_results = results[results["personId"] == person_id]
person_results = person_results.merge(
    competitions[['id', 'name', 'year', 'month', 'day']], how='left',
    left_on='competitionId', right_on='id')

display(person_results.head())

event_results = person_results[person_results['eventId'] == event][[
    'competitionId',
    'roundTypeId',
    'best', 'average', 'personName', 'personId', 'value1', 'value2', 'value3', 'value4',
    'value5', 'year', 'month', 'day'
]].assign(date=lambda x: pd.to_datetime(x[['year', 'month', 'day']]).dt.date).drop(
    columns=['year', 'month', 'day'])

value_columns = [col for col in event_results.columns if col.startswith("value")]
id_vars = [col for col in event_results.columns if col not in value_columns]

event_results = pd.melt(
    event_results,
    id_vars=id_vars,
    value_vars=value_columns,
    var_name="solve_number",
    value_name="result"
)

event_results['date'] = pd.to_datetime(event_results['date'])
greatest_date = event_results['date'].max()

event_results['time_difference'] = (greatest_date.year - event_results[
    'date'].dt.year) * 12 + (greatest_date.month - event_results['date'].dt.month)

dnf_rate = event_results[(event_results.time_difference <= 12) & (event_results.roundTypeId == 'f')]
dnf_rate = dnf_rate[dnf_rate.result == -1].shape[0] / dnf_rate.shape[0]

event_results = event_results[event_results.result != -1]

event_results['weight'] = 2 ** (-event_results['time_difference'] / halflife)

event_results.sort_values(['date'], ascending=False).head(25)

data = event_results['result']
weights = event_results['weight']

# Resample the data based on weights
resampled_data = np.random.choice(data, size=1000, p=weights / weights.sum())

# Fit a Skew Normal distribution
skew_params = skewnorm.fit(resampled_data)

# Fit a Normal distribution
normal_params = norm.fit(resampled_data)

# Calculate MLE for both distributions
mle_skew = calculate_mle(resampled_data, skewnorm, skew_params)
mle_normal = calculate_mle(resampled_data, norm, normal_params)

        
shape, loc, scale = skew_params
delta = shape / np.sqrt(1 + shape**2)
skew_mean = loc + scale * delta * np.sqrt(2 / np.pi)


# Compare MLE values
print("Fitted Skew Normal Parameters (shape, loc, scale):", skew_params)
print("Fitted Normal Parameters (mean, std):", normal_params)
print(f"Skew Normal Mean: {skew_mean:.2f}")
print(f"Skew Normal Log-Likelihood: {mle_skew:.2f}")
print(f"Normal Log-Likelihood: {mle_normal:.2f}")

# Step 3: Visualize the fit
x = np.linspace(resampled_data.min(), resampled_data.max(), 100)
pdf = skewnorm.pdf(x, *skew_params)
norm_pdf = norm.pdf(x, *normal_params)

plt.hist(resampled_data, bins=50, density=True, alpha=0.6, color='gray', label='Resampled Data')
plt.plot(x, pdf, 'r-', label='Fitted Skew Normal PDF')
plt.plot(x, norm_pdf, 'b--', label='Fitted Normal PDF')
plt.title("Fitted Skew Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.show()
```
